```python
import pandas as pd

CSV_FILE = "results/xauusd-m1-performance-test-after-01-01-2022.csv"

performance_test_results_df = pd.read_csv(CSV_FILE).dropna()
performance_test_results_df = performance_test_results_df[performance_test_results_df["rr"] != 1]
performance_test_results_df.set_index(["rl", "rr", "ct"], inplace=True)
performance_test_results_df.sort_values("Sharp ratio", inplace=True, ascending=False)
performance_test_results_df.to_html(CSV_FILE.replace("csv", "html"))
performance_test_results_df.head(20)
```
