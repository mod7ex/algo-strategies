```python
CSV_FILE = "XAUUSD-5M-RB-performance-test.csv"

results_df = pd.read_csv(CSV_FILE).dropna()
results_df = results_df[
    (results_df["ct"] < "17:00") &
    (results_df["rr"] != 1)
]
results_df.set_index(["ct", "rl", "rr", "mlh"], inplace=True)
results_df.sort_values("Max loss streak", inplace=True, ascending=True)
# results_df.sort_values("Sharp ratio", inplace=True, ascending=False)
# results_df.to_html(CSV_FILE.replace("csv", "html"))
results_df.head(20)
```
