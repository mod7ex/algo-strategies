```python
from utils import get_data, run_range_breakout, backtest_stats, print_stats
```

```python
df = get_data("../data/XAUUSD_M5_01-01-2025_to_24-08-2026.csv")
```

```python
MAX_RANGE_DELTA = 3
RANGE_LENGTH = 2
RRR = 4
MAX_LOOK_AHEAD = 15
```

```python
_result = run_range_breakout(
    df,
    range_length=RANGE_LENGTH,
    max_range_delta=MAX_RANGE_DELTA,
    rrr=RRR,
    max_lookahead=MAX_LOOK_AHEAD,
)
```

```python
import plotly.graph_objects as go

fig = go.Figure()

fig.add_trace(go.Scatter(
    x=_result.index,
    y=_result["Balance"],
    mode="lines",
    name="Balance",
    line=dict(
        color="#60A5FA",
        width=3,
        shape="spline"
    ),
    hovertemplate="Balance: %{y:,.2f}<extra></extra>"
))

fig.update_layout(
    template="plotly_dark",
    paper_bgcolor="#08111F",
    plot_bgcolor="#08111F",
    font=dict(color="#CBD5E1"),

    xaxis=dict(
        showgrid=False,
        showline=False
    ),

    yaxis=dict(
        title=None,
        showgrid=True,
        gridcolor="rgba(148,163,184,0.12)",
        tickformat=",.0f"
    ),

    hovermode="x",
    margin=dict(l=20, r=20, t=20, b=20)
)

fig.show()
```

```python
_stats = backtest_stats(_result)
```

```python
print_stats(_stats)
```
